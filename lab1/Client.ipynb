# Running server and client in separate notebooks - client part

Here we start the client. The server needs to be started first. Lets go....

```python
import clientserver # import from our script

client = clientserver.Client() # Create a client instance

client.call("Test") # start the call and print echo
client.get("Jamie") # Get Jamie's phone number
client.get("Hannah") # Get Hannah's phone number
client.get("Bec") # Get Bec's phone number
client.get("Unknown") # Get an unknown name
client.getAll() # Get all entries
client.close()
```
