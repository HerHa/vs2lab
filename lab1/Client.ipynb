# client part

Start the client and Test Echo Service

```python
import clientserver # import from our script

client = clientserver.Client() # Create a client instance

client.call("Test")
client.call("Lets go!")
```

Get phone number for spcific name

```python
client.get("Jamie")
client.get("Hannah")
client.get("Bec")
client.get("Unknown")
```

Get all Phone Numbers

```python
client.getAll()
```

Close Connection

```python
client.close()
```
