# Auskunft Client

Dieses Notebook zeigt die Nutzung des Auskunft-Dienstes.

Der Client verbindet sich über TCP mit dem Server
und ruft die API-Funktionen GET und GETALL auf.
Oder führt die Default echo-Funktion aus.

```python
import clientserver

client = clientserver.Client()

client.call()
```

## GET Beispiel

Abfrage eines einzelnen Eintrags aus der Telefon-Datenbank.

Eingabe: Name

Ausgabe: OK name Telefonnr END


```python
client = clientserver.Client()

client.get("Alice")
```

## GETALL Beispiel

Abfrage aller Einträge aus der Datenbank.

Ausgabe: OK Name Telefonnr Name Telefonnr ... END

```python
client = clientserver.Client()

client.getall()
```
