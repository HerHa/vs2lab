# Auskunft Server

Dieses Notebook startet den Auskunft-Server. 
Er muss *vor* dem Client gestartet werden.
Der Server stellt zwei Operationen bereit:

- GET <Name>
- GETALL

Die Kommunikation erfolgt über TCP-Sockets.

```python
import clientserver # import from our script

server = clientserver.Server() # Create a server instance

server.serve() # start server routine in the forked process
```

## Hinweis

Der Server läuft nun dauerhaft in dieser Zelle.
Er kann nur durch Stoppen der Zelle beendet werden.

Der Server verarbeitet eingehende TCP-Anfragen vom Client.
